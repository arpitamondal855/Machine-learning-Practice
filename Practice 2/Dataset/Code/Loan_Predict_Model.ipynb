```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

#Calling the dataset

```python
data=pd.read_csv("/content/loan_prediction_dataset.csv")
data.head()
```

#Summary of dataset

```python
data.shape
data.info()
data.describe()
data.isnull()
```

#Drop unwanted colums

```python
x=data.drop(columns=['Age','Employment_Status'])
x
```

#Configure the feature and target

```python
x=data[['Income','Credit_Score','Loan_Amount','Loan_Term']]
y=data['Loan_Approved']
```

##Train test split and shape verify

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
x_test.shape
```

```python
y_train.shape
```

##Train logistic regression model

```python
from sklearn.linear_model import LogisticRegression
test=LogisticRegression()
test.fit(x_train,y_train)
```

##Model prediction

```python
predicted=test.predict(x_test)
predicted
```

##Actual vs Prediction

```python
Result=pd.DataFrame({
    "Actual":y_test.values,
    "Predicted":predicted
})
Result
```

##Evaluate model accuracy score

```python
from sklearn.metrics import accuracy_score
acc=accuracy_score(y_test,predicted)
print("Accuracy is :",acc*100,"%")
```

##Confusion matrix and visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

# Configure subplot grid
fig, axes = plt.subplots(1, 2, figsize=(13, 6))

# 1. Confusion Matrix Heatmap
cm = confusion_matrix(y_test, predicted)
sns.heatmap(cm, annot=True, fmt='d', cmap='Reds',
            xticklabels=['Denied (0)', 'Approved (1)'],
            yticklabels=['Denied (0)', 'Approved (1)'], ax=axes[0])
axes[0].set_title('Confusion Matrix Heatmap')
axes[0].set_xlabel('Predicted Label')
axes[0].set_ylabel('Actual Label')

# 2. Comparing Counts of Actual vs Predicted
comparison_df = pd.DataFrame({
    'Value': list(y_test.values) + list(predicted),
    'Type': ['Actual'] * len(y_test) + ['Predicted'] * len(predicted)
})

sns.countplot(x='Value', hue='Type', data=comparison_df, palette='Set2', ax=axes[1])
axes[1].set_title('Distribution of Actual vs. Predicted Loan Status')
axes[1].set_xlabel('Loan Status (0 = Denied, 1 = Approved)')
axes[1].set_ylabel('Count')

plt.tight_layout()
plt.show()
```
