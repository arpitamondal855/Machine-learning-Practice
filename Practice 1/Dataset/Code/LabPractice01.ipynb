##Libraries

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import accuracy_score
```

##importing dataset

```python
df = pd.read_csv('/content/diabetes_012_health_indicators_BRFSS2015.csv')
df.head()
```

##checking informations

```python
df.shape
df.info()
df.describe()
df.isnull()
```

##data separating (train,test)

```python
y = df['Diabetes_012']

x = df.drop('Diabetes_012', axis=1)
x.describe()
```

##Split

```python
x_train,x_test, y_train, y_test = train_test_split(x, y, random_state=1)
```

##Logistic model

```python
diabetes_model = LogisticRegression(max_iter=1000)
diabetes_model.fit(x_train, y_train)
```

##Prediction score

```python
predictions = diabetes_model.predict(x_test)
test_mae = mean_absolute_error(y_test, predictions)
print('Test score: {}'.format(test_mae))
```
